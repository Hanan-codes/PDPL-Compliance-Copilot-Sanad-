```python
from pathlib import Path

project_dir = Path("/content/gap_app/gap_app")

keywords = [
    "STAGE_DATA_DIR",
    "stage1_outputs",
    "stage2_outputs",
    "LLM_PROVIDER",
    "GEMINI_API_KEY",
    "OPENAI_API_KEY",
    "load_engine",
    "st.set_page_config",
]

for filename in ["app.py", "gap_analyser.py", "rag_inference.py"]:
    path = project_dir / filename
    print(f"\n{'=' * 15} {filename} {'=' * 15}")

    if not path.exists():
        print("File not found")
        continue

    lines = path.read_text(encoding="utf-8", errors="replace").splitlines()
    matches = [
        (i, line.strip())
        for i, line in enumerate(lines, 1)
        if any(keyword.lower() in line.lower() for keyword in keywords)
    ]

    for i, line in matches:
        print(f"{i}: {line}")
```

```python
import subprocess
import time
import requests

OLLAMA_URL = "http://127.0.0.1:11434"

# تحقق من الخادم، وشغّله فقط إذا لم يكن يعمل
try:
    r = requests.get(f"{OLLAMA_URL}/api/tags", timeout=3)
    r.raise_for_status()
    print("Ollama server is already running.")
except requests.RequestException:
    ollama_log = open("/tmp/ollama.log", "w")
    ollama_process = subprocess.Popen(
        ["ollama", "serve"],
        stdout=ollama_log,
        stderr=subprocess.STDOUT
    )

    for _ in range(30):
        try:
            r = requests.get(f"{OLLAMA_URL}/api/tags", timeout=3)
            r.raise_for_status()
            print("Ollama server is ready.")
            break
        except requests.RequestException:
            time.sleep(2)
    else:
        raise RuntimeError("Ollama server failed to start. Check /tmp/ollama.log")

# اختبر النموذج
response = requests.post(
    f"{OLLAMA_URL}/api/generate",
    json={
        "model": "qwen2.5:7b",
        "prompt": "أجب بجملة عربية قصيرة: ما الهدف من نظام حماية البيانات الشخصية؟",
        "stream": False
    },
    timeout=240
)
response.raise_for_status()

print("\nModel response:")
print(response.json().get("response", "No response returned"))
```

```python
%cd /content/gap_app/gap_app

import os
import subprocess
import time
from pathlib import Path
import requests

project_dir = Path("/content/gap_app/gap_app")
data_dir = project_dir / "data"

# التأكد من وجود الملفات
assert (project_dir / "app.py").exists(), "app.py غير موجود"
assert (data_dir / "stage1_outputs").exists(), "stage1_outputs غير موجود"
assert (data_dir / "stage2_outputs").exists(), "stage2_outputs غير موجود"

# إعداد Ollama ومسار البيانات
os.environ["LLM_PROVIDER"] = "ollama"
os.environ["OLLAMA_MODEL"] = "qwen2.5:7b"
os.environ["STAGE_DATA_DIR"] = str(data_dir)

# التأكد من أن Ollama يعمل
r = requests.get("http://127.0.0.1:11434/api/tags", timeout=10)
r.raise_for_status()
print("Ollama يعمل.")

# تشغيل Streamlit إذا لم يكن يعمل بالفعل
try:
    r = requests.get(
        "http://127.0.0.1:8501/_stcore/health",
        timeout=3
    )
    if r.ok:
        print("Streamlit يعمل بالفعل.")
    else:
        raise requests.RequestException()
except requests.RequestException:
    streamlit_log = open("/tmp/streamlit.log", "w")

    streamlit_process = subprocess.Popen(
        [
            "streamlit", "run", "app.py",
            "--server.port", "8501",
            "--server.address", "0.0.0.0",
            "--server.headless", "true"
        ],
        cwd=str(project_dir),
        env=os.environ.copy(),
        stdout=streamlit_log,
        stderr=subprocess.STDOUT
    )

    for _ in range(15):
        time.sleep(2)
        if streamlit_process.poll() is not None:
            print(Path("/tmp/streamlit.log").read_text(errors="replace")[-8000:])
            raise RuntimeError("تعذّر تشغيل Streamlit.")

        try:
            r = requests.get(
                "http://127.0.0.1:8501/_stcore/health",
                timeout=2
            )
            if r.ok:
                print("تم تشغيل Streamlit بنجاح.")
                break
        except requests.RequestException:
            pass
    else:
        print(Path("/tmp/streamlit.log").read_text(errors="replace")[-8000:])
        raise RuntimeError("لم يستجب Streamlit خلال الوقت المحدد.")
```

```python

import os
import subprocess
import time
import requests
from pathlib import Path

project_dir = "/content/gap_app/gap_app"
data_dir = f"{project_dir}/data"

# Stop the existing Streamlit process on port 8501
subprocess.run(
    ["bash", "-lc", "fuser -k 8501/tcp"],
    capture_output=True,
    text=True
)
time.sleep(2)

# Configure the application
env = os.environ.copy()
env["LLM_PROVIDER"] = "ollama"
env["OLLAMA_MODEL"] = "qwen2.5:7b"
env["STAGE_DATA_DIR"] = data_dir

# Allow the Colab proxy host and origin
log = open("/tmp/streamlit.log", "w")

streamlit_process = subprocess.Popen(
    [
        "streamlit", "run", "app.py",
        "--server.port=8501",
        "--server.address=0.0.0.0",
        "--server.headless=true",
        "--server.enableCORS=false",
        "--server.enableXsrfProtection=false",
        "--browser.gatherUsageStats=false",
    ],
    cwd=project_dir,
    env=env,
    stdout=log,
    stderr=subprocess.STDOUT,
)

# Wait for the app to become healthy
for _ in range(20):
    time.sleep(2)
    try:
        r = requests.get(
            "http://127.0.0.1:8501/_stcore/health",
            timeout=3
        )
        if r.status_code == 200:
            print("Streamlit is running:", r.text)
            break
    except requests.RequestException:
        pass
else:
    print(Path("/tmp/streamlit.log").read_text(errors="replace")[-8000:])
    raise RuntimeError("Streamlit failed to start.")
```
